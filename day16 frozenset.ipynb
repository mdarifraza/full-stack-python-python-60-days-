"'#frozenset():-
#syntax:-----1.empty frozenset:--varname=frozenset()
            2.non-empty frozenset:---varname=frozenset([val1,val2,....valn])
                                      varname=frozenset((val1,val2,...valn))
                                      varname=frozenset({val1,val2,...valn})"'

```python
fs=frozenset()
print(fs,type(fs),len(fs))
```

```python
fs=frozenset([10,20,10,10,10])
print(fs,type(fs),len(fs))
```

```python
fs=frozenset((10,20,10,10,10))
print(fs,type(fs),len(fs))
```

```python
fs=frozenset({10,20,10,10,10})
print(fs,type(fs),len(fs))
```

```python
fs=frozenset([10,"travis",45.67,True])
print(fs,type(fs),len(fs))
```

```python
#predefined function:
fs1=frozenset({10,20,30,409})
print(fs1,type(fs1),id(fs1),len(fs1))
```

```python
fs2=fs1.copy()
print(fs2,type(fs2),id(fs2),len(fs2))
```
