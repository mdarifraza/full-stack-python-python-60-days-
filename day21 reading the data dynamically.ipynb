### program calculate area and perimeter of rectangle

```python
l=float(input("enter length:"))
b=float(input("enter breadth:"))
ar=l*b
pr=2*(l+b)
print("="*40)
print("\t length={}".format(l))
print("\t breadh={}".format(b))
print("\t area of rectangle={}".format(ar))
print("\t perimeter of recttange={}".format(pr))
```

### program for calculate area of circle:-

```python
r=float(input("enter radius:"))
ac=3.14*r*r
print("="*50)
print("radius:{}".format(r))
print("area of circle:%0.2f"%ac)
print("="*50)
```

## program for calculate area of triangle

```python
b=float(input("enter base:"))
h=float(input("enter height:"))
at=1/2*b*h
print("="*50)
print("\t base:{}".format(b))
print("\t height:{}".format(h))
print("\t area of triangle:{}".format(at))
print("="*50)
```

### program for simple interest and total amount to pay:-

```python
p=float(input("enter principle amount:"))
r=float(input("enter rate of interest:"))
t=float(input("enter time:"))
si=(p*r*t)/100
totamt=p+si
print("="*100)
print("SIMPLE INTEREST CALCULATION")
print("="*100)
print("\t principle amount:{}".format(p))
print("\t rate of interest:{}".format(r))
print("\t time:{}".format(t))
print("\t simple interest:{}".format(si))
print("="*50)
print("TOTAL AMOUNT TO PAY CALCULATION")
print("="*50)
print("\t total amount to pay:{}".format(totamt))
print("="*100)
```
