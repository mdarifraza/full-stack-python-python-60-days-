Dict in Dict

```python
d1={"sno":10,"sname":"rossum","imarks":{"cm":17,"cppm":16,"pym":18},"emarks":{"cm":67,"cppm":78,"pym":70},"cname":"oucet"}
print(d1,type(d1))
```

```python
for k in d1.keys():
    print(k)
```

```python
for v in d1.values():
    print(v)
```

```python
for kv in d1.items():
    print(kv)
```

```python
for k in d1.keys():
    print(k,"--->",d1.get(k))
```

```python
for v in d1.get("imarks"):
    print(v)
```

```python
for v in d1.get("imarks"):
    print(v,"--->",d1.get("imarks").get(v))
```

```python
for v in d1.get("imarks"):
    print(v,">>>>>",d1["imarks"][v])
```

inside of dict,we define list tuple and set also

```python
d1={"sno":10,"sname":"rossum","imarks":[16,18,17],"emarks":(60,77,67),"crs":{"b.tech(cse)","m.tech(ai)"},"cname":"oucet"}
print(d1,type(d1))
```

```python
print(type(d1["imarks"]))
```

```python
print(d1["crs"],type(d1["crs"]))
```

```python
for k,v in d1.items():
    print(k,"--->>",v,">>>>>",type(v))
```
