LIST CATEGORY DATA TYPES (Collection or Data structure):-
Two data types in list category they are:-
1.List(mutable)
2.Tuple(immutable)

```python
#list:-the purpose of list data type is that "To Store multiple values either of same type or different type or both
#the types in single object with unique and dupicate values
```

```python
l1=[10,20,30,40,50,10,40,50,20,10]
print(l1,type(l1))
```

```python
l2=[20,"rossum",34.56,True,2+3j]
print(l2,type(l2))
print(len(l2))
print(l2[0])
print(l2[1])
print(l2[1:4])
```

```python
l2=[20,"rossum",34.56,True,2+3.5j]
print(l2,type(l2),id(l2))
l2[0]=30
print(l2,type(l2),id(l2))
```

```python
l2[2:4]=[44.55,False]
print(l2,type(l2),type(l2))
```

```python
l3=[30,"rossum",34.56,True,(2+3.5j)]
print(l3,type(l3),id(l3))
l3[2:4]=[44.5,False]
print(l3,type(l3),id(l3))
```
