# syntax 4:-
- print(messeges cum values with format())
- or
- print(values cum messeges with format())

```python
a=10
print("val of a={}".format(a))
```

```python
sno=10
name="ram"
print("my number is {}".format(sno),"my name is {}".format(name))
print("my number is {} and my name is {}".format(sno,name))
```

# syntax 5:-
- print(messeges cum values with format specifier)
- or
- print(values cum messeges with format specifier)

```python
sno=10
sname="ram"
marks=44.56
print("my number is %d and my name is %s and marks is %f" %(sno,sname,marks))
```

```python
print("my number is %d and my name is %s and marks is %0.2f" %(sno,sname,marks))
```

# syntax 6:-
- print(value,end=" ")

```python
r=range(10,101,20)
for val in r:
    print(val)
```

```python
for val in r:
    print(val,end=" ")
for val in r:
    print( val,end="---->")
```
