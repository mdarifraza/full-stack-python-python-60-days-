```python
'''6.keys()
syntax:-   varname=dictobj.keys()
                or
            dictobj.keys()
'''
d1={10:"python",20:"java",30:"c++",40:"c"}
print(d1,id(d1))
```

```python
ks=d1.keys()
print(ks,type(ks))
```

```python
for k in ks:
    print(k)
```

```python
for k in d1.keys():
    print(k)
```

```python
for k in d1.keys():
    print(k,"-->",d1.get(k))
```

```python
for k in d1.keys():
    print(k,"--->",d1[k])
```

```python
'''7.values()
syntax:-     varname=dictobj.values()
                    or
            dictobj.values()
'''
d1={10:"python",20:"java",30:"c++",40:"c"}
print(d1,id(d1))
```

```python
vs=d1.values()
```

```python
'''8.items()
syntax:-   varname=d1.items()
'''
d1={10:"python",20:"java",30:"c++",40:"c"}
print(d1)
```
